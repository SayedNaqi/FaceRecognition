```python

# ============================================================
# FACE RECOGNITION PROJECT
# Environment Test + Image Face Detection + Real-Time Webcam
# Python 3.12 | OpenCV | NumPy | Matplotlib | Pillow
# ============================================================

# -------------------------
# 1. IMPORT LIBRARIES
# -------------------------

import sys
import time
import cv2
import numpy as np
import matplotlib
import matplotlib.pyplot as plt

from PIL import Image
from pathlib import Path


# -------------------------
# 2. PROJECT SETTINGS
# -------------------------

# Set True to detect faces from a saved image
RUN_IMAGE_DETECTION = False

# Set True to start real-time webcam detection
RUN_WEBCAM = True

# Image location
IMAGE_NAME = "test.jpg"

# Webcam camera index
CAMERA_INDEX = 0


# -------------------------
# 3. ENVIRONMENT INFORMATION
# -------------------------

print("=" * 55)
print("       FACE RECOGNITION PROJECT")
print("=" * 55)

print("Python Version :", sys.version)
print("Python Path    :", sys.executable)
print("OpenCV Version :", cv2.__version__)
print("NumPy Version  :", np.__version__)
print("Matplotlib     :", matplotlib.__version__)

print("\nEnvironment check completed!")


# -------------------------
# 4. LOAD FACE DETECTION MODEL
# -------------------------

cascade_path = (
    cv2.data.haarcascades
    + "haarcascade_frontalface_default.xml"
)

face_detector = cv2.CascadeClassifier(cascade_path)

if face_detector.empty():
    raise RuntimeError(
        "Face Detection Model could not be loaded!"
    )

print("\nFace Detection Model: READY")


# -------------------------
# 5. IMAGE FACE DETECTION
# -------------------------

def detect_faces_in_image(image_path):

    image_path = Path(image_path)

    if not image_path.exists():
        raise FileNotFoundError(
            f"Image not found: {image_path}"
        )

    # Read image
    image = cv2.imread(str(image_path))

    if image is None:
        raise ValueError(
            "Could not read the image!"
        )

    # Convert BGR to grayscale
    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    # Detect faces
    faces = face_detector.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    # Copy original image
    result = image.copy()

    # Draw rectangle around detected faces
    for (x, y, w, h) in faces:

        cv2.rectangle(
            result,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            3
        )

        cv2.putText(
            result,
            "Face Detected",
            (x, max(y - 10, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )

    # Convert BGR to RGB
    result_rgb = cv2.cvtColor(
        result,
        cv2.COLOR_BGR2RGB
    )

    # Display result
    plt.figure(figsize=(12, 8))
    plt.imshow(result_rgb)
    plt.title(
        f"Face Detection | Faces Found: {len(faces)}"
    )
    plt.axis("off")
    plt.show()

    print("\nImage:", image_path)
    print("Image dimensions:", image.shape)
    print("Number of detected faces:", len(faces))

    return faces, result


# -------------------------
# 6. FIND PROJECT IMAGE
# -------------------------

def find_project_image():

    current_dir = Path.cwd()

    possible_paths = [
        current_dir / "data" / "images" / IMAGE_NAME,
        current_dir.parent / "data" / "images" / IMAGE_NAME,
        current_dir / IMAGE_NAME
    ]

    for path in possible_paths:

        if path.exists():
            return path

    raise FileNotFoundError(
        "\nImage not found!\n"
        "Please put your image here:\n"
        "FaceRecognition/data/images/test.jpg"
    )


# -------------------------
# 7. RUN IMAGE DETECTION
# -------------------------

if RUN_IMAGE_DETECTION:

    print("\n" + "=" * 55)
    print("          IMAGE FACE DETECTION")
    print("=" * 55)

    image_path = find_project_image()

    detected_faces, image_result = (
        detect_faces_in_image(image_path)
    )


# -------------------------
# 8. REAL-TIME WEBCAM DETECTION
# -------------------------

def run_webcam_detection():

    print("\n" + "=" * 55)
    print("          REAL-TIME FACE DETECTION")
    print("=" * 55)

    # Open webcam
    camera = cv2.VideoCapture(CAMERA_INDEX)

    if not camera.isOpened():
        raise RuntimeError(
            "Cannot access webcam!\n"
            "Check camera connection and permissions."
        )

    print("Webcam connected successfully!")
    print("Press Q to stop the webcam.")

    frame_count = 0
    start_time = time.time()

    try:

        while True:

            # Read webcam frame
            success, frame = camera.read()

            if not success:
                print("Cannot read webcam frame!")
                break

            # Convert frame to grayscale
            gray = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2GRAY
            )

            # Detect faces
            faces = face_detector.detectMultiScale(
                gray,
                scaleFactor=1.1,
                minNeighbors=5,
                minSize=(40, 40)
            )

            # Draw rectangles
            for (x, y, w, h) in faces:

                cv2.rectangle(
                    frame,
                    (x, y),
                    (x + w, y + h),
                    (0, 255, 0),
                    2
                )

                cv2.putText(
                    frame,
                    "Face Detected",
                    (x, max(y - 10, 20)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.7,
                    (0, 255, 0),
                    2
                )

            # Calculate FPS
            frame_count += 1

            elapsed = time.time() - start_time

            fps = (
                frame_count / elapsed
                if elapsed > 0
                else 0
            )

            # Display information
            cv2.putText(
                frame,
                f"Faces: {len(faces)} | FPS: {fps:.1f}",
                (10, 30),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 0, 255),
                2
            )

            # Show webcam window
            cv2.imshow(
                "Real-Time Face Detection",
                frame
            )

            # Press Q to exit
            if cv2.waitKey(1) & 0xFF == ord("q"):
                break

    finally:

        camera.release()
        cv2.destroyAllWindows()

        print("Webcam stopped!")


# -------------------------
# 9. RUN WEBCAM IF ENABLED
# -------------------------

if RUN_WEBCAM:

    run_webcam_detection()


# -------------------------
# 10. PROJECT COMPLETION
# -------------------------

print("\n" + "=" * 55)
print("       FACE DETECTION PROCESS COMPLETED")
print("=" * 55)

print("Image Detection:", RUN_IMAGE_DETECTION)
print("Webcam Detection:", RUN_WEBCAM)

print("\nNext Step: Modern Face Detection & Recognition")
```
